### Descriptive summary Statistics


```python
# Descriptive summary statistics provide an overview of studentس scores.
# It includes measures such as mean, minimum, maximum, and standard deviation.
# These statistics help us understand the overall performance of students.
# We can also identify if there are any unusually high or low scores.

df.describe()
```

###Distribution of Math Score

```python
# Distribution of Math Score, which represents students performance in mathematics.
# The histogram shows that most students scored around the average range,
# while a few students achieved very high or very low scores.

plt.figure(figsize=(8,5))
sns.histplot(df['math score'], bins=20)
plt.title("Distribution of Math Score")
plt.show()
```

###Distribution of Reading Score


```python
# Distribution of Reading Score, which represents students performance in reading.
# The histogram shows that scores are relatively spread,
# with most students performing within a middle range.

plt.figure(figsize=(8,5))
sns.histplot(df['reading score'], bins=20)
plt.title("Distribution of Reading Score")
plt.show()



```

###Distribution of Writing Score

```python
# Distribution of Writing Score, which represents students performance in writing.
# The histogram shows that most students scored moderately,
# with fewer students at the extreme high or low ends.

plt.figure(figsize=(8,5))
sns.histplot(df['writing score'], bins=20)
plt.title("Distribution of Writing Score")
plt.show()
```



###Average Score by Gender

```python
# Average scores based on gender.
# The bar chart shows differences in performance between male and female students,
# which helps identify any performance gaps.

gender_scores = df.groupby('gender')[['math score','reading score','writing score']].mean()

gender_scores.plot(kind='bar', figsize=(8,5))
plt.title("Average Scores by Gender")
plt.ylabel("Score")
plt.show()

```

### Average Math Score by Parental Education


```python
# Average Math Score by Parental Level of Education.
# This analysis shows how parents' education level affects students' performance in math.
# The bar chart indicates that students with more educated parents tend to have higher average math scores.
# This suggests that parental education may play an important role in academic achievement.

education_score = df.groupby('parental level of education')['math score'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,5))
education_score.plot(kind='bar')
plt.title("Average Math Score by Parental Education")
plt.ylabel("Average Math Score")
plt.show()

education_score

```

### Top Average Scores by Gender




```python
# Average Math Score by Gender.
# This analysis compares the average math scores between male and female students.
# The result shows which gender performs better on average in mathematics.
# This helps identify any performance gap between genders.

gender_scores = df.groupby('gender')['math score'].mean().sort_values(ascending=False)

gender_scores
```

### Math vs Reading Score Relationship




```python
# Relationship between Math Score and Reading Score.
# The scatter plot shows how students performance in math relates to their performance in reading.
# There appears to be a positive relationship, where students with higher math scores
# also tend to have higher reading scores.
# This indicates that strong academic performance in one subject is often associated with others.

plt.figure(figsize=(8,5))
sns.scatterplot(x='math score', y='reading score', data=df)
plt.title("Math Score vs Reading Score")
plt.show()

```

###Math score distribution by test prepration cource

```python
# Score by Test Preparation Course, which shows how completing the preparation course
# affects students' math performance.
# The boxplot highlights differences in score distribution between the two groups.

plt.figure(figsize=(10,5))
sns.boxplot(x='test preparation course', y='math score', data=df)
plt.title("Math Score Distribution by Test Preparation Course")
plt.show()
```

### Correlation Matrix

```python
# Correlation matrix showing relationships between scores.
# The heatmap indicates strong positive correlations between all subjects,
# especially between reading and writing scores.

plt.figure(figsize=(6,4))
sns.heatmap(df[['math score','reading score','writing score']].corr(), annot=True)
plt.title("Correlation Matrix")
plt.show()

```

- Correlation close to 1 = strong positive relationship
- Close to 0 = weak relationship


### Scores Distribution by Gender





```python
# Score by Gender, which shows the distribution of math scores for male and female students.
# The boxplot displays the median, spread, and potential outliers for each group.
# It helps compare performance differences and variability between genders.

plt.figure(figsize=(8,5))
sns.boxplot(x='gender', y='math score', data=df)
plt.title("Math Scores Distribution by Gender")
plt.show()
```
